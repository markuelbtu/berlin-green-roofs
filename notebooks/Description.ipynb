# Project Context and Pipeline Overview

The workspace contains a central `data` directory with the following structure:
- **Raw input layers**: `Bezirk_Neukoelln`, `green_roofs_berlin`, `solarpotential_berlin`, `solarpotential_roofsegments_neukoelln`
- **Processed outputs**: `exports` (all intermediate and final results are written here)
- **Get data here**: [click here!] (https://tubcloud.tu-berlin.de/s/aHDHTq29pB6pZJH)
### Notebook Pipeline

### preprocessing_step0_gr_roof_cl

This notebook performs the initial cleaning of Berlin-wide green-roof building data.
It removes underground parking structures (`Tiefgarage`) to prevent misleading roof classifications in later ML steps.

**Main input**
- `data/green_roofs_berlin/buildings_berlin.shp`

**Main outputs**
- Folder: `data/exports/processing_step0_gr_roof_cl/`
- `buildings_cleaned_no_tiefgarage.csv`
- `buildings_cleaned_no_tiefgarage.gpkg`
- `buildings_cleaned_no_tiefgarage.shp`

### preprocessing_step1_clip_to_neuk

This notebook clips city-wide datasets to the district of Berlin-Neukoelln to reduce data volume and focus the analysis area.
It also prepares a model-ready roof dataset and a clipped solar potential layer.

**Main inputs**
- `data/exports/processing_step0_gr_roof_cl/buildings_cleaned_no_tiefgarage.gpkg`
- `data/Bezirk_Neukoelln/Bezirk_Neukoelln.shp`
- `data/solarpotential_berlin/solarpotential.shp`

**Main outputs**
- Folder: `data/exports/preprocessing_step1_clip_to_neuk/`
- `neukoelln_greenroofs.csv`
- `neukoelln_greenroofs.gpkg`
- `neukoelln_greenroofs.shp`
- `Solarpotential_Neukoelln.gpkg`
- `Solarpotential_Neukoelln.shp`

### preprocessing_step2_sol_suit

This notebook combines roof, solar potential, and roof-segment slope data to compute a suitability indicator.
The key rule is a slope threshold of 15 degrees, used to estimate suitable roof area per building.

**Main inputs**
- `data/exports/preprocessing_step1_clip_to_neuk/neukoelln_greenroofs.shp`
- `data/exports/preprocessing_step1_clip_to_neuk/Solarpotential_Neukoelln.shp`
- `data/solarpotential_roofsegments_neukoelln/roof_segments_Neukoelln.shp`

**Main outputs**
- Folder: `data/exports/preprocessing_step2_sol_suit/`
- `solar_segment_suitability.csv`
- `preprocessed_buildings_sol_suit.csv` 
- `preprocessed_buildings_sol_suit.gpkg`
- `preprocessed_buildings_sol_suit.shp`

### preprocessing_step3_geosp_join

This notebook performs the core geospatial join needed for machine learning.
It joins Neukoelln roof polygons with enriched building-level suitability attributes and prefixes building fields with `bldg_` for consistency.

**Main inputs**
- `data/exports/preprocessing_step1_clip_to_neuk/neukoelln_greenroofs.shp`
- `data/exports/preprocessing_step2_sol_suit/preprocessed_buildings_sol_suit.shp`

**Main outputs**
- Folder: `data/exports/preprocessing_step3_geosp_join/`
- `geospatial_join_complete_neukoelln.csv`
- `geospatial_join_complete_neukoelln.gpkg`
- `geospatial_join_complete_neukoelln.shp`

### DecisionTree_RandomForest_pred

This notebook trains and evaluates the prediction models (Decision Tree and Random Forest), including threshold tuning.
It also creates final suitability classes for green roof upgrade potential.

**Main input**
- `data/exports/preprocessing_step3_geosp_join/geospatial_join_complete_neukoelln.csv`

**Main outputs**
- Folder: `data/exports/solution/`
- `roof_suitability_predictions.csv`
- `roof_suitability_predictions.gpkg`
- `roof_suitability_predictions.shp`

**Final suitability classes**
- `not appropriate`
- `medium appropriate`
- `highly appropriate`
